## Define features and target

```python
y = df["Value"]

X = df.drop(columns=[
    "Name", "ID", "Nationality", "BP", "foot", "Joined",
    "Value", "Wage", "Release Clause"
])
print("Features & Target Prepared")
```

## Regression Test

```python
def regression_fitness_test(X, y):
    model = LinearRegression()
    scores = cross_val_score(model, X, y, cv=5, scoring="r2")
    
    print("\nRegression Fitness Test")
    print("------------------------")
    print("R² scores per fold:", scores)
    print("Mean R²:", np.mean(scores))
    
    if np.mean(scores) > 0.5:
        print("Dataset is FIT for regression.")
    else:
        print("Dataset is NOT suitable for regression.")

regression_fitness_test(X, y)
```
